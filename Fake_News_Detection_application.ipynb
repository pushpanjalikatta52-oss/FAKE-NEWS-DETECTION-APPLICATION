```python
# ============================================================
# FAKE NEWS DETECTION APPLICATION
# Google Colab - Single Cell
# ============================================================

!pip -q install transformers torch gradio

import torch
import gradio as gr
from transformers import pipeline

# ------------------------------------------------------------
# Load Fake News Detection Model
# ------------------------------------------------------------

print("Loading Fake News Detection model...")

classifier = pipeline(
    "text-classification",
    model="hamzab/roberta-fake-news-classification",
    tokenizer="hamzab/roberta-fake-news-classification"
)

print("Model loaded successfully!")


# ------------------------------------------------------------
# Fake News Detection Function
# ------------------------------------------------------------

def detect_fake_news(text):

    if not text.strip():
        return "Please enter a news article or headline."

    try:
        result = classifier(
            text[:2000],
            truncation=True
        )[0]

        label = result["label"]
        score = result["score"] * 100

        # Convert model labels into readable output
        if "fake" in label.lower():
            prediction = "🚨 FAKE NEWS"
        else:
            prediction = "✅ REAL NEWS"

        return f"""
{prediction}

Confidence: {score:.2f}%

Model Label: {label}
"""

    except Exception as e:
        return f"Error: {str(e)}"


# ------------------------------------------------------------
# Gradio Application
# ------------------------------------------------------------

with gr.Blocks(title="Fake News Detection") as app:

    gr.Markdown(
        """
        # 📰 Fake News Detection Application

        ### AI-Based News Classification

        Enter a news headline or article below and the AI model
        will classify it as **Real News** or **Fake News**.
        """
    )

    news_input = gr.Textbox(
        label="Enter News Text",
        placeholder="Paste a news headline or article here...",
        lines=10
    )

    detect_button = gr.Button(
        "🔍 Detect News",
        variant="primary"
    )

    result_output = gr.Textbox(
        label="Detection Result",
        lines=6
    )

    detect_button.click(
        fn=detect_fake_news,
        inputs=news_input,
        outputs=result_output
    )

    gr.Markdown(
        """
        **Technologies:** Python | NLP | Transformers | RoBERTa | Gradio | Google Colab
        """
    )


# ------------------------------------------------------------
# Launch Application
# ------------------------------------------------------------

app.launch(share=True)
```
